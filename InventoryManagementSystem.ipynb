### Inventory Management System

```python
class InventoryManagementSystem:
    def __init__(self):
        self.inventory = {}

    # Add Product
    def add_product(self):
        product_id = input("Enter Product ID: ")

        if product_id in self.inventory:
            print("Product ID already exists!")
            return

        name = input("Enter Product Name: ")
        quantity = int(input("Enter Quantity: "))
        price = float(input("Enter Price: "))

        self.inventory[product_id] = {
            "name": name,
            "quantity": quantity,
            "price": price
        }

        print("Product added successfully!")

    # Update Product
    def update_product(self):
        product_id = input("Enter Product ID to update: ")

        if product_id not in self.inventory:
            print("Product not found!")
            return

        print("Leave blank to keep existing value.")

        name = input(f"Name ({self.inventory[product_id]['name']}): ")
        quantity = input(f"Quantity ({self.inventory[product_id]['quantity']}): ")
        price = input(f"Price ({self.inventory[product_id]['price']}): ")

        if name:
            self.inventory[product_id]["name"] = name
        if quantity:
            self.inventory[product_id]["quantity"] = int(quantity)
        if price:
            self.inventory[product_id]["price"] = float(price)

        print("Product updated successfully!")

    # Delete Product
    def delete_product(self):
        product_id = input("Enter Product ID to delete: ")

        if product_id in self.inventory:
            del self.inventory[product_id]
            print("Product deleted successfully!")
        else:
            print("Product not found!")

    # Search Product
    def search_product(self):
        product_id = input("Enter Product ID: ")

        if product_id in self.inventory:
            product = self.inventory[product_id]
            print("\nProduct Details")
            print("----------------")
            print("ID:", product_id)
            print("Name:", product["name"])
            print("Quantity:", product["quantity"])
            print("Price:", product["price"])
        else:
            print("Product not found!")

    # Display All Products
    def display_products(self):
        if not self.inventory:
            print("Inventory is empty!")
            return

        print("\nInventory")
        print("-" * 60)
        print(f"{'ID':<10}{'Name':<20}{'Qty':<10}{'Price':<10}")
        print("-" * 60)

        for pid, product in self.inventory.items():
            print(f"{pid:<10}{product['name']:<20}{product['quantity']:<10}{product['price']:<10.2f}")

    # Inventory Report
    def inventory_report(self):
        total_products = len(self.inventory)
        total_quantity = sum(item["quantity"] for item in self.inventory.values())
        total_value = sum(item["quantity"] * item["price"] for item in self.inventory.values())

        print("\nInventory Report")
        print("----------------------")
        print("Total Products :", total_products)
        print("Total Quantity :", total_quantity)
        print(f"Total Inventory Value : ${total_value:.2f}")


def main():
    ims = InventoryManagementSystem()

    while True:
        print("\n===== Inventory Management System =====")
        print("1. Add Product")
        print("2. Update Product")
        print("3. Delete Product")
        print("4. Search Product")
        print("5. Display Products")
        print("6. Inventory Report")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            ims.add_product()
        elif choice == "2":
            ims.update_product()
        elif choice == "3":
            ims.delete_product()
        elif choice == "4":
            ims.search_product()
        elif choice == "5":
            ims.display_products()
        elif choice == "6":
            ims.inventory_report()
        elif choice == "7":
            print("Thank you!")
            break
        else:
            print("Invalid choice! Please try again.")


if __name__ == "__main__":
    main()
```
